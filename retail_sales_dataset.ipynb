```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
path=r"/content/archive (12).zip"
data=pd.read_csv(path)
print(data.shape)
print(data.head())
```

```python
print(data.isnull().sum())
print(data.dtypes)
```

```python
objects_data=data.select_dtypes(include=["object"]).columns
data[objects_data]=data[objects_data].fillna("Missing")



print(data.isnull().sum())
```

```python
# Get all float64 columns
float_data = data.select_dtypes(include=["float64"]).columns

# Fill NaNs in each float column with its respective column mean
data[float_data] = data[float_data].fillna(data[float_data].mean())

# Verify if any null values remain
print(data.isnull().sum())
```
